```python
# ==========================================================
# FULL CNN + GAT + CAPSULE NETWORK + ADAPTIVE TEMPORAL ATTENTION
# FOR TON-IoT DATASET WITH MULTI-AGENT FEDERATED LEARNING
# AND TRUST-AWARE AGGREGATION (5 AGENTS)
#
# + EXPLAINABLE AI:
#    * Integrated Gradients  -> feature-level attribution
#    * GNNExplainer          -> explanatory subgraphs / communication paths
#    * Combined contribution summary (High / Medium / Low)
#
# Requirements:
#   pip install tensorflow scikit-learn imbalanced-learn pandas numpy matplotlib seaborn networkx
# ==========================================================

# ==========================================================
# 0. IMPORTS & GLOBAL SETUP
# ==========================================================
import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.metrics import classification_report, confusion_matrix
from tensorflow.keras.models import Model, load_model
from tensorflow.keras.layers import (Input, Conv1D, Dense, Dropout,
                                     GlobalAveragePooling1D, LayerNormalization,
                                     MultiHeadAttention, Lambda)
from tensorflow.keras.callbacks import ReduceLROnPlateau, ModelCheckpoint
from tensorflow.keras.utils import to_categorical
from imblearn.over_sampling import SMOTE

SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)

# Grow GPU memory lazily (avoids OOM on shared GPUs / Kaggle)
for _gpu in tf.config.list_physical_devices('GPU'):
    try:
        tf.config.experimental.set_memory_growth(_gpu, True)
    except Exception:
        pass

print("TensorFlow:", tf.__version__)
print("GPU devices:", tf.config.list_physical_devices('GPU'))

# ---------------- RUN CONFIGURATION (reduce for quick runs) ----------------
NUM_AGENTS   = 5
NUM_ROUNDS   = 20
LOCAL_EPOCHS = 20
BATCH_SIZE   = 128
COMMUNICATION_TOPOLOGY = 'ring'          # Peer-to-peer topology

# Trust aggregation hyperparameters
ALPHA = 0.4                              # weight for update consistency
BETA  = 0.3                              # weight for loss behavior
GAMMA = 0.3                              # weight for historical reliability
TRUST_HISTORY_DECAY = 0.9                # decay factor for historical trust update

# XAI configuration
N_EXPLAIN     = 500    # test samples explained by Integrated Gradients
IG_STEPS      = 64     # Riemann-sum steps for IG (higher = more accurate)
GNN_PER_CLASS = 3      # samples per class explained by GNNExplainer
GNN_EPOCHS    = 200    # optimization steps per GNNExplainer sample

# ==========================================================
# 1. CUSTOM LAYERS
# ==========================================================
class AdaptiveTemporalAttention(tf.keras.layers.Layer):
    """
    Applies adaptive attention across the feature dimension (temporal axis).
    Input: (batch, num_features, filters)
    Output: (batch, num_features, filters) with each feature scaled.
    """
    def __init__(self, reduction=16, **kwargs):
        super(AdaptiveTemporalAttention, self).__init__(**kwargs)
        self.reduction = reduction

    def build(self, input_shape):
        self.num_features = input_shape[1]
        self.filters = input_shape[2]
        self.dense1 = Dense(self.num_features // self.reduction, activation='relu')
        self.dense2 = Dense(self.num_features, activation='sigmoid')
        super().build(input_shape)

    def call(self, inputs):
        pooled = tf.reduce_mean(inputs, axis=-1)
        scores = self.dense1(pooled)
        scores = self.dense2(scores)
        scores = tf.expand_dims(scores, axis=-1)
        return inputs * scores

    def get_config(self):
        config = super().get_config()
        config.update({'reduction': self.reduction})
        return config


class NodeToClassCaps(tf.keras.layers.Layer):
    """
    Takes GAT node outputs as Primary Capsules and routes them to Class Capsules.
    Input: (batch, num_nodes, node_dim) -> Output: (batch, num_classes, capsule_dim)
    """
    def __init__(self, num_classes, capsule_dim, num_routing=3, **kwargs):
        super(NodeToClassCaps, self).__init__(**kwargs)
        self.num_classes = num_classes
        self.capsule_dim = capsule_dim
        self.num_routing = num_routing

    def build(self, input_shape):
        self.num_nodes = input_shape[1]
        self.node_dim = input_shape[2]
        self.W = self.add_weight(
            shape=(self.num_classes, self.num_nodes, self.node_dim, self.capsule_dim),
            initializer='glorot_uniform',
            trainable=True,
            name='W'
        )
        super(NodeToClassCaps, self).build(input_shape)

    def call(self, inputs):
        batch_size = tf.shape(inputs)[0]
        inputs = self.squash(inputs)
        W_perm = tf.transpose(self.W, perm=[1, 0, 2, 3])
        u_hat = tf.einsum('b i d, i j d a -> b j i a', inputs, W_perm)
        b = tf.zeros((batch_size, self.num_classes, self.num_nodes))
        for r in range(self.num_routing):
            c = tf.nn.softmax(b, axis=1)
            s = tf.reduce_sum(tf.expand_dims(c, axis=-1) * u_hat, axis=2)
            v = self.squash(s)
            if r < self.num_routing - 1:
                b += tf.reduce_sum(tf.expand_dims(v, axis=2) * u_hat, axis=-1)
        return v

    def squash(self, vector):
        vec_squared_norm = tf.reduce_sum(tf.square(vector), axis=-1, keepdims=True)
        scalar_factor = vec_squared_norm / (1.0 + vec_squared_norm)
        return scalar_factor * vector / (tf.sqrt(vec_squared_norm + tf.keras.backend.epsilon()))

    def compute_output_shape(self, input_shape):
        return (input_shape[0], self.num_classes, self.capsule_dim)

    def get_config(self):
        config = super().get_config()
        config.update({
            'num_classes': self.num_classes,
            'capsule_dim': self.capsule_dim,
            'num_routing': self.num_routing,
        })
        return config

# ==========================================================
# 2. LOSS / OUTPUT UTILS (module-level so they can be serialized)
# ==========================================================
def margin_loss(y_true, y_pred):
    """Margin loss for capsule networks."""
    L = y_true * tf.square(tf.maximum(0., 0.9 - y_pred)) + \
        0.5 * (1 - y_true) * tf.square(tf.maximum(0., y_pred - 0.1))
    return tf.reduce_mean(tf.reduce_sum(L, axis=1))


def capsule_length(vector):
    """Compute the length (norm) of capsule vectors."""
    return tf.sqrt(tf.reduce_sum(tf.square(vector), axis=-1) + tf.keras.backend.epsilon())

# ==========================================================
# 3. GRAPH CONSTRUCTION
# ==========================================================
def build_feature_correlation_graph(X_train_np, threshold=0.2):
    corr_matrix = np.corrcoef(X_train_np, rowvar=False)
    A = (np.abs(corr_matrix) > threshold).astype(np.float32)
    np.fill_diagonal(A, 1.0)
    mask = np.where(A > 0, 0.0, -1e9)   # 0 = keep edge, -1e9 = block edge
    return mask

# ==========================================================
# 4. DATA LOADING & PREPROCESSING
# ==========================================================
df = pd.read_csv('/kaggle/input/datasets/arnobbhowmik/ton-iot-network-dataset/train_test_network.csv')
print(f"Original shape: {df.shape}")

y = df['type']
drop_cols = ['dns_query', 'ssl_subject', 'ssl_issuer', 'http_uri', 'http_user_agent',
             'http_orig_mime_types', 'http_resp_mime_types', 'weird_addl', 'type']
if 'label' in df.columns:
    drop_cols.append('label')
drop_cols = [col for col in drop_cols if col in df.columns]
X = df.drop(columns=drop_cols)

X = X.replace('-', np.nan)
for col in X.select_dtypes(include=['int64', 'float64']).columns:
    X[col] = X[col].fillna(X[col].median())
for col in X.select_dtypes(include=['object']).columns:
    X[col] = X[col].fillna('Unknown')
    X[col] = LabelEncoder().fit_transform(X[col].astype(str))

target_encoder = LabelEncoder()
y_encoded = target_encoder.fit_transform(y)
num_classes = len(target_encoder.classes_)
y_encoded = to_categorical(y_encoded, num_classes)
print(f"Features used: {X.shape[1]} | Classes: {num_classes} -> {list(target_encoder.classes_)}")

X_train, X_test, y_train, y_test = train_test_split(
    X, y_encoded, test_size=0.2, stratify=y, random_state=SEED)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

y_train_int = np.argmax(y_train, axis=1)
X_train_res, y_train_int = SMOTE(random_state=SEED, k_neighbors=5).fit_resample(X_train_scaled, y_train_int)
y_train = to_categorical(y_train_int, num_classes)

num_features = X_train_res.shape[1]
gat_mask = build_feature_correlation_graph(X_train_res, threshold=0.2)

X_train_final = X_train_res.reshape((-1, num_features, 1))
X_test_final = X_test_scaled.reshape((-1, num_features, 1))
print(f"Train shape after SMOTE: {X_train_final.shape} | Test shape: {X_test_final.shape}")

# ==========================================================
# 5. MODEL CREATION FUNCTION
# ==========================================================
def create_hybrid_model():
    """Creates and compiles the hybrid CNN+GAT+CapsNet model."""
    inputs = Input(shape=(num_features, 1))

    # 1. CNN BRANCH
    x = Conv1D(filters=64, kernel_size=3, padding='same', activation='relu')(inputs)
    x = Conv1D(filters=64, kernel_size=3, padding='same', activation='relu')(x)
    cnn_out = x

    # 2. ADAPTIVE TEMPORAL ATTENTION
    x = AdaptiveTemporalAttention(reduction=16)(x)

    # 3. GAT BRANCH (feature-correlation graph as attention mask)
    x = MultiHeadAttention(num_heads=4, key_dim=16, attention_axes=1)(
        query=x, key=x, value=x, attention_mask=gat_mask
    )
    x = LayerNormalization()(x + cnn_out)

    # 4. CAPSULE NETWORK BRANCH
    digit_caps = NodeToClassCaps(
        num_classes=num_classes,
        capsule_dim=16,
        num_routing=3
    )(x)

    outputs = Lambda(capsule_length)(digit_caps)

    model = Model(inputs=inputs, outputs=outputs)
    model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
                  loss=margin_loss, metrics=['accuracy'])
    return model

# Create the global model (server)
global_model = create_hybrid_model()
global_model.summary()

# ==========================================================
# 6. MULTI-AGENT FEDERATED LEARNING SETUP
# ==========================================================
# Split training data among agents
num_train_samples = X_train_final.shape[0]
indices = np.random.permutation(num_train_samples)
agent_data_indices = np.array_split(indices, NUM_AGENTS)

agent_datasets = []
for idx in agent_data_indices:
    agent_X = X_train_final[idx]
    agent_y = y_train[idx]
    agent_datasets.append((agent_X, agent_y))

# Initialize global weights
global_weights = global_model.get_weights()

# Create local models for each agent
agent_models = [create_hybrid_model() for _ in range(NUM_AGENTS)]

# Historical trust scores (initialized equal)
hist_trust = np.ones(NUM_AGENTS) / NUM_AGENTS

# ==========================================================
# 7. FEDERATED TRAINING WITH TRUST-AWARE AGGREGATION
# ==========================================================
best_val_acc = 0.0
best_weights_snapshot = None
history_federated = {'val_accuracy': [], 'val_loss': [], 'trust_scores': []}

for round_num in range(1, NUM_ROUNDS + 1):
    print(f"\n--- Trust-Aware Multi-Agent Federated Round {round_num}/{NUM_ROUNDS} ---")

    # Store global weights before local training (for update computation)
    global_weights_before = [np.copy(w) for w in global_weights]

    # Local training on each agent and collect info for trust computation
    local_weights_list = []
    local_losses = []
    update_vectors_flat = []   # flattened update vector per agent

    for agent_id in range(NUM_AGENTS):
        print(f"  Agent {agent_id+1}/{NUM_AGENTS} local training...")
        agent_X, agent_y = agent_datasets[agent_id]

        # Start from current global weights
        agent_models[agent_id].set_weights(global_weights_before)

        # Local training
        history_local = agent_models[agent_id].fit(
            agent_X, agent_y,
            epochs=LOCAL_EPOCHS,
            batch_size=BATCH_SIZE,
            verbose=0
        )

        # Collect training loss (last epoch)
        train_loss = history_local.history['loss'][-1]
        local_losses.append(train_loss)

        # Get updated weights and compute update vector
        agent_weights = agent_models[agent_id].get_weights()
        local_weights_list.append(agent_weights)

        # Flatten the difference (update) into a single vector
        update_flat = np.concatenate([(w - gw).flatten()
                                      for w, gw in zip(agent_weights, global_weights_before)])
        update_vectors_flat.append(update_flat)

    # ----- Compute Trust Scores -----
    # 1. Update Consistency (cosine similarity to average update)
    avg_update = np.mean(update_vectors_flat, axis=0)
    consistency_scores = []
    for vec in update_vectors_flat:
        dot = np.dot(vec, avg_update)
        norm_vec = np.linalg.norm(vec)
        norm_avg = np.linalg.norm(avg_update)
        cos_sim = dot / (norm_vec * norm_avg + 1e-8)
        # Shift cosine from [-1,1] to [0,1]
        consistency_scores.append((cos_sim + 1) / 2.0)
    consistency_scores = np.array(consistency_scores)

    # 2. Loss Behavior (inverse of loss, normalized)
    losses = np.array(local_losses)
    inv_losses = 1.0 / (losses + 1e-8)
    loss_scores = (inv_losses - inv_losses.min()) / (inv_losses.max() - inv_losses.min() + 1e-8)

    # 3. Historical Reliability (already stored from previous rounds)
    hist_scores = hist_trust.copy()

    # Combine scores
    raw_trust = ALPHA * consistency_scores + BETA * loss_scores + GAMMA * hist_scores

    # Normalize trust scores to sum to 1
    trust_scores = raw_trust / (np.sum(raw_trust) + 1e-8)

    # Update historical trust (exponential moving average)
    hist_trust = TRUST_HISTORY_DECAY * hist_trust + (1 - TRUST_HISTORY_DECAY) * trust_scores

    print(f"  Trust scores (normalized): {np.round(trust_scores, 4)}")

    # ----- Peer-to-Peer Communication (Ring Topology) -----
    print("  Performing peer-to-peer communication (ring topology)...")
    communicated_weights = [w.copy() for w in local_weights_list]

    if COMMUNICATION_TOPOLOGY == 'ring':
        for i in range(NUM_AGENTS):
            neighbor = (i + 1) % NUM_AGENTS
            avg_pair = []
            for w_i, w_neighbor in zip(local_weights_list[i], local_weights_list[neighbor]):
                avg_pair.append((w_i + w_neighbor) / 2.0)
            communicated_weights[i] = avg_pair

    # ----- Trust-Aware Server Aggregation (weighted average) -----
    avg_weights = []
    for weight_tuple in zip(*communicated_weights):
        weighted_avg = np.average(weight_tuple, axis=0, weights=trust_scores)
        avg_weights.append(weighted_avg)

    # Update global model
    global_weights = avg_weights
    global_model.set_weights(global_weights)

    # Evaluate global model on test set
    val_loss, val_acc = global_model.evaluate(X_test_final, y_test, verbose=0)
    history_federated['val_loss'].append(val_loss)
    history_federated['val_accuracy'].append(val_acc)
    history_federated['trust_scores'].append(trust_scores.copy())
    print(f"  Round {round_num} - Global model val_accuracy: {val_acc:.4f}, val_loss: {val_loss:.4f}")

    # Save best model (on disk AND in memory, so XAI never fails on load issues)
    if val_acc > best_val_acc:
        best_val_acc = val_acc
        best_weights_snapshot = [np.copy(w) for w in global_weights]
        global_model.save('best_trust_aware_federated_model.keras')
        print(f"  New best model saved (val_acc = {best_val_acc:.4f})")

print("\nTrust-Aware Multi-Agent Federated training completed.")
print(f"Best validation accuracy: {best_val_acc:.4f}")

# ==========================================================
# 8. FINAL EVALUATION
# ==========================================================
custom_objects = {
    'NodeToClassCaps': NodeToClassCaps,
    'margin_loss': margin_loss,
    'capsule_length': capsule_length,
    'AdaptiveTemporalAttention': AdaptiveTemporalAttention
}

try:
    best_model = load_model('best_trust_aware_federated_model.keras',
                            custom_objects=custom_objects)
    print("Loaded best model from disk.")
except Exception as e:
    print(f"Could not load saved model ({e}); using in-memory best weights instead.")
    best_model = global_model
    if best_weights_snapshot is not None:
        best_model.set_weights(best_weights_snapshot)

test_loss, test_acc = best_model.evaluate(X_test_final, y_test, verbose=0)
print(f"\nBest Trust-Aware Federated Hybrid Model Test Accuracy: {test_acc:.4f}")

y_pred = best_model.predict(X_test_final, verbose=0)
y_pred_classes = np.argmax(y_pred, axis=1)
y_true_classes = np.argmax(y_test, axis=1)

print("\n" + "="*60)
print("CLASSIFICATION REPORT (Trust-Aware Federated CNN + GAT + CapsNet + Adaptive Temporal Attention)")
print("="*60)
print(classification_report(y_true_classes, y_pred_classes, target_names=target_encoder.classes_))

# -------------------- CONFUSION MATRICES --------------------
def plot_confusion_matrix(y_true, y_pred, target_names, title='Confusion Matrix', normalize=False):
    cm = confusion_matrix(y_true, y_pred)
    fmt = '.2%' if normalize else 'd'
    plt.figure(figsize=(10, 8))
    sns.heatmap(cm.astype('float') / cm.sum(axis=1)[:, np.newaxis] if normalize else cm,
                annot=True, fmt=fmt, cmap='Blues',
                xticklabels=target_names, yticklabels=target_names)
    plt.title(title, fontsize=14, fontweight='bold')
    plt.ylabel('True Label', fontsize=12)
    plt.xlabel('Predicted Label', fontsize=12)
    plt.tight_layout()
    plt.show()

plot_confusion_matrix(y_true_classes, y_pred_classes, target_encoder.classes_,
                      title='Trust-Aware Federated Hybrid Model (Raw Counts)')
plot_confusion_matrix(y_true_classes, y_pred_classes, target_encoder.classes_,
                      title='Trust-Aware Federated Hybrid Model (Normalized)', normalize=True)

# -------------------- FEDERATED HISTORY PLOTS --------------------
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.plot(history_federated['val_accuracy'], label='Validation Accuracy')
plt.title('Trust-Aware FL Validation Accuracy per Round')
plt.xlabel('Round'); plt.ylabel('Accuracy'); plt.legend()
plt.subplot(1, 2, 2)
plt.plot(history_federated['val_loss'], label='Validation Loss')
plt.title('Trust-Aware FL Validation Loss per Round')
plt.xlabel('Round'); plt.ylabel('Loss'); plt.legend()
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 5))
trust_array = np.array(history_federated['trust_scores'])
for i in range(NUM_AGENTS):
    plt.plot(trust_array[:, i], label=f'Agent {i+1}')
plt.title('Trust Score Evolution per Round')
plt.xlabel('Round'); plt.ylabel('Trust Score')
plt.legend(); plt.grid(True)
plt.show()

# ==========================================================
# 9. XAI — SHARED SETUP
# ==========================================================
feature_names = list(X.columns)
num_features  = len(feature_names)
w             = max(len(n) for n in feature_names)   # column width for printing
rng           = np.random.default_rng(SEED)

def contribution_label(v, hi=0.66, mid=0.33):
    if v >= hi:  return 'High'
    if v >= mid: return 'Medium'
    return 'Low'

# ==========================================================
# 10. XAI 1 — INTEGRATED GRADIENTS (feature-level attribution)
# ==========================================================
def integrated_gradients(model, X_samples, target_classes=None,
                         baseline=None, steps=IG_STEPS, batch_size=32):
    """
    IG(x) = (x - baseline) * mean_alpha[ dF(baseline + alpha*(x-baseline)) / dx ]
    Default baseline = zeros (= feature mean for StandardScaler data).
    Returns attributions (N, F, 1) and the explained class per sample (N,).
    """
    N, F, _ = X_samples.shape
    if baseline is None:
        baseline = np.zeros((F, 1), dtype=np.float32)
    baseline = np.broadcast_to(baseline.astype(np.float32), (N, F, 1))

    if target_classes is None:
        target_classes = np.argmax(model.predict(X_samples, verbose=0), axis=1)
    target_classes = np.asarray(target_classes)

    alphas = np.linspace(0.0, 1.0, steps, dtype=np.float32)
    attributions = np.zeros((N, F, 1), dtype=np.float32)

    for s in range(0, N, batch_size):
        e  = min(s + batch_size, N)
        xb = X_samples[s:e].astype(np.float32)
        bb = baseline[s:e]
        tb = target_classes[s:e]
        B  = e - s

        # (steps, B, F, 1) path interpolation
        inter = bb[None] + alphas[:, None, None, None] * (xb[None] - bb[None])
        inter = inter.reshape(steps * B, F, 1)

        grads = np.zeros_like(inter)
        tgt_flat = np.tile(tb, steps)                 # flat idx = step*B + b
        for gs in range(0, inter.shape[0], batch_size):
            ge = min(gs + batch_size, inter.shape[0])
            xi = tf.convert_to_tensor(inter[gs:ge])
            with tf.GradientTape() as tape:
                tape.watch(xi)
                out = model(xi, training=False)       # capsule lengths
                picked = tf.gather(out, tgt_flat[gs:ge], batch_dims=1)
                score = tf.reduce_sum(picked)
            g = tape.gradient(score, xi)
            grads[gs:ge] = (tf.zeros_like(xi) if g is None else g).numpy()

        grads = grads.reshape(steps, B, F, 1).mean(axis=0)   # avg path gradient
        attributions[s:e] = (xb - bb) * grads

    return attributions, target_classes

# ---- stratified subset of the test set ----
explain_idx = []
for c in np.unique(y_true_classes):
    idx_c = np.where(y_true_classes == c)[0]
    n_c   = min(len(idx_c), max(1, int(N_EXPLAIN * len(idx_c) / len(y_true_classes))))
    explain_idx += list(rng.choice(idx_c, size=n_c, replace=False))
explain_idx = np.array(explain_idx)

print(f"\nExplaining {len(explain_idx)} test samples with Integrated Gradients...")
X_explain = X_test_final[explain_idx]
y_explain = y_true_classes[explain_idx]

ig_attrs, ig_targets = integrated_gradients(best_model, X_explain,
                                            target_classes=y_explain)

# ---- global feature importance ----
global_ig = np.abs(ig_attrs).mean(axis=0).squeeze(-1)
ig_norm   = global_ig / (global_ig.max() + 1e-12)
order     = np.argsort(global_ig)[::-1]

print("\n" + "="*60)
print("INTEGRATED GRADIENTS — FEATURE-LEVEL ATTRIBUTION")
print("="*60)
for f in order:
    print(f"{feature_names[f]:<{w}} → {contribution_label(ig_norm[f]):<6} "
          f"contribution  (score={global_ig[f]:.4f}, norm={ig_norm[f]:.2f})")

# ---- per-class importance ----
print("\nTop-5 features per class (mean |IG|):")
per_class_ig = {}
for c in range(num_classes):
    m = ig_targets == c
    if m.sum() == 0:
        continue
    per_class_ig[c] = np.abs(ig_attrs[m]).mean(axis=0).squeeze(-1)
    top = np.argsort(per_class_ig[c])[::-1][:5]
    print(f"  {target_encoder.classes_[c]:<20}: " +
          ", ".join(f"{feature_names[i]} ({per_class_ig[c][i]:.3f})" for i in top))

top_k = 15
plt.figure(figsize=(10, 6))
plt.barh([feature_names[i] for i in order[:top_k]][::-1],
         global_ig[order[:top_k]][::-1], color='steelblue')
plt.title(f'Integrated Gradients — Top {top_k} Global Feature Attributions')
plt.xlabel('Mean |attribution|')
plt.tight_layout(); plt.show()

# ==========================================================
# 11. XAI 2 — GNNEXPLAINER (subgraph + communication-path attribution)
# Learns soft edge masks over the feature-correlation graph and a
# node/feature mask by maximising fidelity of the original prediction
# under the masked graph (original GNNExplainer objective).
# ==========================================================

# Rebuild adjacency A (build_feature_correlation_graph only returned the mask)
corr_matrix = np.nan_to_num(np.corrcoef(X_train_res, rowvar=False), nan=0.0)
A_adj = (np.abs(corr_matrix) > 0.2).astype(np.float32)
np.fill_diagonal(A_adj, 1.0)


class GNNExplainerTF:
    BLOCK_VALUE = -10.0   # soft "edge removed" bias (keeps gradients alive)

    def __init__(self, model, A, base_mask, feature_names,
                 epochs=GNN_EPOCHS, lr=0.05, lambda_sparsity=1e-3, lambda_entropy=1e-3):
        self.model, self.feature_names = model, feature_names
        self.A, self.A_np = tf.constant(A, tf.float32), A
        self.base_mask = tf.constant(base_mask, tf.float32)   # 0=edge, -1e9=none
        self.F = len(feature_names)
        self.epochs, self.lr = epochs, lr
        self.lambda_sparsity, self.lambda_entropy = lambda_sparsity, lambda_entropy

        # --- decompose the trained model ---
        self.gat_layer = next(l for l in model.layers if isinstance(l, MultiHeadAttention))
        self.ln_layer  = next(l for l in model.layers if isinstance(l, LayerNormalization))
        conv_out = [l for l in model.layers if isinstance(l, Conv1D)][-1].output
        attn_out = next(l for l in model.layers
                        if isinstance(l, AdaptiveTemporalAttention)).output
        self.pre_gat = Model(model.input, [conv_out, attn_out])   # -> (cnn_out, gat_in)
        self.head    = Model(self.ln_layer.output, model.output)  # post-GAT -> capsule

    def _masked_forward(self, x, edge_logits, feat_logits):
        feat_mask = tf.sigmoid(feat_logits)                # (F,) node mask
        xm = x * feat_mask[None, :, None]                  # mask input features
        cnn_out, H = self.pre_gat(xm, training=False)

        keep = tf.sigmoid(edge_logits)                     # (F,F) edge importance
        soft_block = self.A * (1.0 - keep) * self.BLOCK_VALUE
        eff_mask = self.base_mask + soft_block             # non-edges stay blocked

        a = self.gat_layer(H, H, H, attention_mask=eff_mask, training=False)
        z = self.ln_layer(a + cnn_out, training=False)
        return self.head(z, training=False)

    def explain(self, x_np, verbose=False):
        """x_np: (F, 1) single scaled sample -> importance dict."""
        x = tf.convert_to_tensor(x_np[None, ...], tf.float32)

        p_orig = self.model(x, training=False)
        target = int(tf.argmax(p_orig[0]))

        edge_logits = tf.Variable(np.full((self.F, self.F), 2.0, np.float32))
        feat_logits = tf.Variable(np.full((self.F,), 2.0, np.float32))
        opt = tf.keras.optimizers.Adam(self.lr)
        eps = 1e-8

        for ep in range(self.epochs):
            with tf.GradientTape() as tape:
                p_mask = self._masked_forward(x, edge_logits, feat_logits)
                fidelity = tf.reduce_sum(tf.square(p_mask - p_orig))  # reproduce decision

                keep  = tf.sigmoid(edge_logits)
                fmask = tf.sigmoid(feat_logits)
                sparsity = (tf.reduce_sum(tf.abs(keep * self.A)) / tf.reduce_sum(self.A)
                            + tf.reduce_mean(tf.abs(fmask)))
                ent_e = -tf.reduce_sum(self.A * (keep * tf.math.log(keep + eps)
                                       + (1 - keep) * tf.math.log(1 - keep + eps))
                                       ) / tf.reduce_sum(self.A)
                ent_f = -tf.reduce_mean(fmask * tf.math.log(fmask + eps)
                                        + (1 - fmask) * tf.math.log(1 - fmask + eps))
                loss = (fidelity + self.lambda_sparsity * sparsity
                        + self.lambda_entropy * (ent_e + ent_f))

            grads = tape.gradient(loss, [edge_logits, feat_logits])
            opt.apply_gradients(zip(grads, [edge_logits, feat_logits]))
            if verbose and (ep + 1) % 50 == 0:
                print(f"    explainer epoch {ep+1:3d} | loss={loss.numpy():.5f}")

        return {'target': target,
                'orig_prob': float(p_orig.numpy()[0, target]),
                'edge_importance': (tf.sigmoid(edge_logits).numpy() * self.A_np).astype(np.float32),
                'node_importance': tf.sigmoid(feat_logits).numpy().astype(np.float32)}


# ---- run on a few samples per class ----
explainer = GNNExplainerTF(best_model, A_adj, gat_mask, feature_names,
                           epochs=GNN_EPOCHS, lr=0.05,
                           lambda_sparsity=1e-3, lambda_entropy=1e-3)

gnn_indices = []
for c in range(num_classes):
    idx_c = np.where(y_explain == c)[0]
    if len(idx_c):
        gnn_indices += list(rng.choice(idx_c, size=min(GNN_PER_CLASS, len(idx_c)), replace=False))

agg_edges = np.zeros((num_features, num_features), np.float32)
agg_nodes = np.zeros(num_features, np.float32)
per_class_edges = {c: [] for c in range(num_classes)}

print(f"\nRunning GNNExplainer on {len(gnn_indices)} samples ({explainer.epochs} epochs each)...")
for n, gi in enumerate(gnn_indices):
    res = explainer.explain(X_test_final[explain_idx[gi]])
    agg_edges += res['edge_importance']
    agg_nodes += res['node_importance']
    per_class_edges[res['target']].append(res['edge_importance'])
    print(f"  [{n+1}/{len(gnn_indices)}] true={target_encoder.classes_[y_explain[gi]]}  "
          f"pred={target_encoder.classes_[res['target']]} (p={res['orig_prob']:.3f})")

agg_edges /= max(1, len(gnn_indices))
agg_nodes /= max(1, len(gnn_indices))

# ---- report top communication paths ----
all_edges = [(i, j, agg_edges[i, j]) for i in range(num_features)
             for j in range(i + 1, num_features) if agg_edges[i, j] > 1e-4]
all_edges.sort(key=lambda t: -t[2])

print("\n" + "="*60)
print("GNNEXPLAINER — TOP 'COMMUNICATION PATHS' (feature–feature edges)")
print("="*60)
for i, j, wt in all_edges[:12]:
    print(f"  {feature_names[i]:<{w}} ↔ {feature_names[j]:<{w}}  importance={wt:.3f}")

print("\nGNNEXPLAINER — TOP NODES (features) in the explanatory subgraph:")
for f in np.argsort(agg_nodes)[::-1][:10]:
    print(f"  {feature_names[f]:<{w}} importance={agg_nodes[f]:.3f}")

print("\nPer-class explanatory subgraphs (top-3 paths per class):")
for c in range(num_classes):
    if not per_class_edges[c]:
        continue
    ce = np.mean(per_class_edges[c], axis=0)
    eds = sorted([(i, j, ce[i, j]) for i in range(num_features)
                  for j in range(i + 1, num_features) if ce[i, j] > 1e-4],
                 key=lambda t: -t[2])
    print(f"  Class '{target_encoder.classes_[c]}':")
    for i, j, wt in eds[:3]:
        print(f"     {feature_names[i]} ↔ {feature_names[j]}  ({wt:.3f})")

# ---- plots ----
plt.figure(figsize=(13, 11))
sns.heatmap(agg_edges, xticklabels=feature_names, yticklabels=feature_names,
            cmap='mako', square=True, cbar_kws={'label': 'Edge importance'})
plt.title('GNNExplainer — Aggregated Edge Importance (explanatory subgraph)')
plt.xticks(rotation=90); plt.yticks(rotation=0)
plt.tight_layout(); plt.show()

try:
    import networkx as nx
    if all_edges:
        G = nx.Graph()
        for i, j, wt in all_edges[:12]:
            G.add_edge(feature_names[i], feature_names[j], weight=wt)
        node_w = {feature_names[f]: agg_nodes[f] for f in range(num_features)}
        plt.figure(figsize=(12, 9))
        pos = nx.spring_layout(G, k=1.3, seed=SEED)
        nx.draw_networkx_edges(G, pos, width=[G[u][v]['weight'] * 6 for u, v in G.edges()],
                               edge_color='steelblue', alpha=0.8)
        nx.draw_networkx_nodes(G, pos,
                               node_size=[150 + 1500 * node_w[n] for n in G.nodes()],
                               node_color='orange', alpha=0.9)
        nx.draw_networkx_labels(G, pos, font_size=9)
        plt.title('GNNExplainer — Important Subgraph (communication paths)')
        plt.axis('off'); plt.tight_layout(); plt.show()
    else:
        print("No significant edges found — skipping graph plot.")
except ImportError:
    print("networkx not installed — skipping graph plot.")

# ==========================================================
# 12. XAI 3 — COMBINED FEATURE CONTRIBUTION SUMMARY
# ==========================================================
gnn_norm       = agg_nodes / (agg_nodes.max() + 1e-12)
combined       = 0.5 * ig_norm + 0.5 * gnn_norm
combined_order = np.argsort(combined)[::-1]

print("\n" + "="*60)
print("FEATURE CONTRIBUTION SUMMARY (Integrated Gradients + GNNExplainer)")
print("="*60)
for f in combined_order:
    print(f"{feature_names[f]:<{w}} → {contribution_label(combined[f]):<6} contribution")

# ---- save XAI artifacts to disk ----
pd.DataFrame({
    'feature': feature_names,
    'ig_mean_abs': global_ig,
    'ig_normalized': ig_norm,
    'gnn_node_importance': agg_nodes,
    'gnn_normalized': gnn_norm,
    'combined_score': combined
}).sort_values('combined_score', ascending=False).to_csv('xai_feature_importance.csv', index=False)

edge_rows = [{'feature_1': feature_names[i], 'feature_2': feature_names[j], 'importance': wt}
             for i, j, wt in all_edges]
pd.DataFrame(edge_rows).to_csv('xai_gnnexplainer_edges.csv', index=False)
print("\nSaved: xai_feature_importance.csv, xai_gnnexplainer_edges.csv")

print("\nAll done.")
```

```python
# ==========================================
# HYBRID CNN + GAT + CAPSULE + ADAPTIVE ATTENTION
# FOR CIC-IoMT 2024 DATASET – NO SMOTE
# ==========================================

import os
import pickle
import joblib
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import classification_report, confusion_matrix
from tensorflow.keras.models import Model, load_model
from tensorflow.keras.layers import (Input, Conv1D, Dense,
                                     LayerNormalization, MultiHeadAttention,
                                     Lambda)
from tensorflow.keras.callbacks import ReduceLROnPlateau, ModelCheckpoint, EarlyStopping
from tensorflow.keras.utils import to_categorical
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns

# -------------------- CACHE SETUP --------------------
CACHE_DIR = "cache_iomt"
os.makedirs(CACHE_DIR, exist_ok=True)

def save_cache(data, name):
    if name.endswith('.npy'):
        np.save(os.path.join(CACHE_DIR, name), data)
    else:
        joblib.dump(data, os.path.join(CACHE_DIR, name))

def load_cache(name):
    path = os.path.join(CACHE_DIR, name)
    if not os.path.exists(path):
        return None
    if name.endswith('.npy'):
        return np.load(path, allow_pickle=True)
    else:
        return joblib.load(path)

# -------------------- CUSTOM ATTENTION LAYER --------------------
class AdaptiveTemporalAttention(tf.keras.layers.Layer):
    def __init__(self, reduction=16, **kwargs):
        super(AdaptiveTemporalAttention, self).__init__(**kwargs)
        self.reduction = reduction

    def build(self, input_shape):
        self.num_features = input_shape[1]
        self.filters = input_shape[2]
        self.dense1 = Dense(self.num_features // self.reduction, activation='relu')
        self.dense2 = Dense(self.num_features, activation='sigmoid')
        super().build(input_shape)

    def call(self, inputs):
        pooled = tf.reduce_mean(inputs, axis=-1)
        scores = self.dense1(pooled)
        scores = self.dense2(scores)
        scores = tf.expand_dims(scores, axis=-1)
        return inputs * scores

    def get_config(self):
        config = super().get_config()
        config.update({'reduction': self.reduction})
        return config

# -------------------- CAPSULE LAYER --------------------
class NodeToClassCaps(tf.keras.layers.Layer):
    def __init__(self, num_classes, capsule_dim, num_routing=3, **kwargs):
        super(NodeToClassCaps, self).__init__(**kwargs)
        self.num_classes = num_classes
        self.capsule_dim = capsule_dim
        self.num_routing = num_routing

    def build(self, input_shape):
        self.num_nodes = input_shape[1]
        self.node_dim = input_shape[2]
        self.W = self.add_weight(
            shape=(self.num_classes, self.num_nodes, self.node_dim, self.capsule_dim),
            initializer='glorot_uniform',
            trainable=True,
            name='W'
        )
        super().build(input_shape)

    def call(self, inputs):
        batch_size = tf.shape(inputs)[0]
        inputs = self.squash(inputs)
        W_perm = tf.transpose(self.W, perm=[1, 0, 2, 3])
        u_hat = tf.einsum('b i d, i j d a -> b j i a', inputs, W_perm)

        b = tf.zeros((batch_size, self.num_classes, self.num_nodes))
        for r in range(self.num_routing):
            c = tf.nn.softmax(b, axis=1)
            s = tf.reduce_sum(tf.expand_dims(c, axis=-1) * u_hat, axis=2)
            v = self.squash(s)
            if r < self.num_routing - 1:
                b += tf.reduce_sum(tf.expand_dims(v, axis=2) * u_hat, axis=-1)
        return v

    def squash(self, vector):
        vec_squared_norm = tf.reduce_sum(tf.square(vector), axis=-1, keepdims=True)
        scalar_factor = vec_squared_norm / (1.0 + vec_squared_norm)
        return scalar_factor * vector / (tf.sqrt(vec_squared_norm + tf.keras.backend.epsilon()))

    def compute_output_shape(self, input_shape):
        return (input_shape[0], self.num_classes, self.capsule_dim)

    def get_config(self):
        config = super().get_config()
        config.update({
            'num_classes': self.num_classes,
            'capsule_dim': self.capsule_dim,
            'num_routing': self.num_routing,
        })
        return config

# -------------------- GRAPH CONSTRUCTION --------------------
def build_feature_correlation_graph(X_train_np, threshold=0.2):
    corr_matrix = np.corrcoef(X_train_np, rowvar=False)
    A = (np.abs(corr_matrix) > threshold).astype(np.float32)
    np.fill_diagonal(A, 1.0)
    mask = np.where(A > 0, 0.0, -1e9)
    return mask

# -------------------- DATA LOADING & CACHING --------------------
train_path = "/kaggle/input/datasets/amineipad/cic-iomt-dataset-2024/train_iomt.csv"
test_path  = "/kaggle/input/datasets/amineipad/cic-iomt-dataset-2024/test_iomt.csv"

cache_ready = (load_cache("X_train_scaled.npy") is not None and
               load_cache("X_test_scaled.npy") is not None and
               load_cache("y_train.npy") is not None and
               load_cache("y_test.npy") is not None and
               load_cache("label_encoder.joblib") is not None and
               load_cache("scaler.joblib") is not None and
               load_cache("feature_encoders.joblib") is not None and
               load_cache("gat_mask.npy") is not None)

if cache_ready:
    print("✅ Loading preprocessed data from cache...")
    X_train_scaled = load_cache("X_train_scaled.npy")
    y_train        = load_cache("y_train.npy")
    X_test_scaled  = load_cache("X_test_scaled.npy")
    y_test         = load_cache("y_test.npy")
    label_encoder  = load_cache("label_encoder.joblib")
    scaler         = load_cache("scaler.joblib")
    gat_mask       = load_cache("gat_mask.npy")
    num_classes    = len(label_encoder.classes_)
    num_features   = X_train_scaled.shape[1]
else:
    print("🔄 Cache not found. Processing from scratch...")

    # Load raw data (pyarrow engine = much faster)
    train_df = pd.read_csv(train_path, engine='pyarrow')
    test_df  = pd.read_csv(test_path,  engine='pyarrow')
    print("Train shape:", train_df.shape)
    print("Test shape:", test_df.shape)

    X_train_raw = train_df.drop("label", axis=1)
    y_train_raw = train_df["label"]
    X_test_raw  = test_df.drop("label", axis=1)
    y_test_raw  = test_df["label"]

    # Remove unseen labels from test
    train_labels_set = set(y_train_raw)
    mask = y_test_raw.isin(train_labels_set)
    if not mask.all():
        print(f"⚠️ Removing {sum(~mask)} test samples with unseen labels.")
        X_test_raw = X_test_raw[mask]
        y_test_raw = y_test_raw[mask]

    # -------- Preprocess features (vectorized, no chained assignment) --------
    def preprocess_dataframe(df, fit_encoders=None):
        df = df.copy()
        df = df.replace('-', np.nan)

        # numeric columns — fill in ONE pass
        num_cols = df.select_dtypes(include=['int64', 'float64']).columns
        df[num_cols] = df[num_cols].fillna(df[num_cols].median())

        # categorical columns
        encoders = {}
        obj_cols = df.select_dtypes(include=['object']).columns
        df[obj_cols] = df[obj_cols].fillna('Unknown')
        for col in obj_cols:
            if fit_encoders is not None and col in fit_encoders:
                le = fit_encoders[col]
                df[col] = le.transform(df[col].astype(str))
            else:
                le = LabelEncoder()
                df[col] = le.fit_transform(df[col].astype(str))
                encoders[col] = le
        return df, encoders

    X_train, feature_encoders = preprocess_dataframe(X_train_raw)
    X_test, _ = preprocess_dataframe(X_test_raw, fit_encoders=feature_encoders)

    # Encode labels
    label_encoder = LabelEncoder()
    y_train = label_encoder.fit_transform(y_train_raw)
    y_test  = label_encoder.transform(y_test_raw)
    num_classes = len(label_encoder.classes_)

    # Scale features
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train).astype(np.float32)
    X_test_scaled  = scaler.transform(X_test).astype(np.float32)

    # Build correlation mask
    gat_mask = build_feature_correlation_graph(X_train_scaled, threshold=0.2)

    # Cache everything
    save_cache(X_train_scaled, "X_train_scaled.npy")
    save_cache(y_train, "y_train.npy")
    save_cache(X_test_scaled, "X_test_scaled.npy")
    save_cache(y_test, "y_test.npy")
    save_cache(label_encoder, "label_encoder.joblib")
    save_cache(scaler, "scaler.joblib")
    save_cache(feature_encoders, "feature_encoders.joblib")
    save_cache(gat_mask, "gat_mask.npy")
    print("✅ Cached preprocessed data.")

num_features = X_train_scaled.shape[1]
num_classes  = len(label_encoder.classes_)
print("Classes:", num_classes, "| Features:", num_features)

# -------------------- CLASS WEIGHTS (replaces SMOTE) --------------------
class_weights_raw = compute_class_weight('balanced',
                                         classes=np.unique(y_train),
                                         y=y_train)
CLASS_WEIGHTS = dict(enumerate(class_weights_raw))
print("Class weights:", CLASS_WEIGHTS)

# Reshape for CNN
X_train_final = X_train_scaled.reshape((-1, num_features, 1))
X_test_final  = X_test_scaled.reshape((-1, num_features, 1))

# Categorical labels (needed for margin loss)
y_train_cat = to_categorical(y_train, num_classes)
y_test_cat  = to_categorical(y_test,  num_classes)

# -------------------- BUILD HYBRID MODEL --------------------
inputs = Input(shape=(num_features, 1))
x = Conv1D(filters=64, kernel_size=3, padding='same', activation='relu')(inputs)
x = Conv1D(filters=64, kernel_size=3, padding='same', activation='relu')(x)
cnn_out = x

x = AdaptiveTemporalAttention(reduction=16)(x)

x = MultiHeadAttention(num_heads=4, key_dim=16, attention_axes=1)(
    query=x, key=x, value=x, attention_mask=gat_mask
)
x = LayerNormalization()(x + cnn_out)
                                                                              .
digit_caps = NodeToClassCaps(num_classes=num_classes, capsule_dim=16, num_routing=3)(x)

def capsule_length(vector):
    return tf.sqrt(tf.reduce_sum(tf.square(vector), axis=-1) + tf.keras.backend.epsilon())

outputs = Lambda(capsule_length)(digit_caps)
model = Model(inputs=inputs, outputs=outputs)

# Per-sample margin loss (required for class_weight to work)
def margin_loss(y_true, y_pred):
    L = y_true * tf.square(tf.maximum(0., 0.9 - y_pred)) + \
        0.5 * (1 - y_true) * tf.square(tf.maximum(0., y_pred - 0.1))
    return tf.reduce_sum(L, axis=1)   # per-sample vector

model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
              loss=margin_loss, metrics=['accuracy'])
model.summary()

# -------------------- TRAINING --------------------
BATCH_SIZE = 1024   # large batch = far fewer steps/epoch

checkpoint = ModelCheckpoint('best_iomt_hybrid.keras',
                             monitor='val_accuracy', mode='max',
                             save_best_only=True, verbose=1)
reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.2,
                              patience=3, min_lr=1e-5, verbose=1)
early_stop = EarlyStopping(monitor='val_accuracy', mode='max',
                           patience=7, restore_best_weights=True, verbose=1)

history = model.fit(X_train_final, y_train_cat,
                    epochs=30,
                    batch_size=BATCH_SIZE,
                    validation_data=(X_test_final, y_test_cat),
                    class_weight=CLASS_WEIGHTS,
                    callbacks=[checkpoint, reduce_lr, early_stop],
                    verbose=1)

model.save('final_iomt_hybrid.keras')

# -------------------- EVALUATION --------------------
custom_objects = {
    'NodeToClassCaps': NodeToClassCaps,
    'margin_loss': margin_loss,
    'capsule_length': capsule_length,
    'AdaptiveTemporalAttention': AdaptiveTemporalAttention
}

best_model = load_model('best_iomt_hybrid.keras', custom_objects=custom_objects)

test_loss, test_acc = best_model.evaluate(X_test_final, y_test_cat, verbose=0)
print(f"\nBest Hybrid Model Test Accuracy: {test_acc:.4f}")

y_pred = best_model.predict(X_test_final, batch_size=2048)
y_pred_classes = np.argmax(y_pred, axis=1)
y_true_classes = y_test   # already integer-encoded

print("\n" + "="*60)
print("CLASSIFICATION REPORT")
print("="*60)
print(classification_report(y_true_classes, y_pred_classes,
                            target_names=label_encoder.classes_))

cm = confusion_matrix(y_true_classes, y_pred_classes)
plt.figure(figsize=(12,8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=label_encoder.classes_,
            yticklabels=label_encoder.classes_)
plt.title("Confusion Matrix – Hybrid Model")
plt.xlabel("Predicted"); plt.ylabel("Actual")
plt.xticks(rotation=45); plt.yticks(rotation=45)
plt.tight_layout()
plt.show()

plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
plt.plot(history.history['accuracy'], label='Train Acc')
plt.plot(history.history['val_accuracy'], label='Val Acc')
plt.title('Model Accuracy'); plt.xlabel('Epoch'); plt.legend()

plt.subplot(1,2,2)
plt.plot(history.history['loss'], label='Train Loss')
plt.plot(history.history['val_loss'], label='Val Loss')
plt.title('Margin Loss'); plt.xlabel('Epoch'); plt.legend()
plt.tight_layout()
plt.show()
```

```python

import os
import pickle
import joblib
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import classification_report, confusion_matrix
from tensorflow.keras.models import Model, load_model
from tensorflow.keras.layers import (Input, Conv1D, Dense,
                                     LayerNormalization, MultiHeadAttention,
                                     Lambda)
from tensorflow.keras.callbacks import ReduceLROnPlateau, ModelCheckpoint, EarlyStopping
from tensorflow.keras.utils import to_categorical
import tensorflow as tf
import matplotlib.pyplot as plt
import seaborn as sns
import random

# -------------------- CACHE SETUP --------------------
CACHE_DIR = "cache_iomt"
os.makedirs(CACHE_DIR, exist_ok=True)

def save_cache(data, name):
    if name.endswith('.npy'):
        np.save(os.path.join(CACHE_DIR, name), data)
    else:
        joblib.dump(data, os.path.join(CACHE_DIR, name))

def load_cache(name):
    path = os.path.join(CACHE_DIR, name)
    if not os.path.exists(path):
        return None
    if name.endswith('.npy'):
        return np.load(path, allow_pickle=True)
    else:
        return joblib.load(path)

# -------------------- CUSTOM ATTENTION LAYER --------------------
class AdaptiveTemporalAttention(tf.keras.layers.Layer):
    def __init__(self, reduction=16, **kwargs):
        super(AdaptiveTemporalAttention, self).__init__(**kwargs)
        self.reduction = reduction

    def build(self, input_shape):
        self.num_features = input_shape[1]
        self.filters = input_shape[2]
        self.dense1 = Dense(self.num_features // self.reduction, activation='relu')
        self.dense2 = Dense(self.num_features, activation='sigmoid')
        super().build(input_shape)

    def call(self, inputs):
        pooled = tf.reduce_mean(inputs, axis=-1)
        scores = self.dense1(pooled)
        scores = self.dense2(scores)
        scores = tf.expand_dims(scores, axis=-1)
        return inputs * scores

    def get_config(self):
        config = super().get_config()
        config.update({'reduction': self.reduction})
        return config

# -------------------- CAPSULE LAYER --------------------
class NodeToClassCaps(tf.keras.layers.Layer):
    def __init__(self, num_classes, capsule_dim, num_routing=3, **kwargs):
        super(NodeToClassCaps, self).__init__(**kwargs)
        self.num_classes = num_classes
        self.capsule_dim = capsule_dim
        self.num_routing = num_routing

    def build(self, input_shape):
        self.num_nodes = input_shape[1]
        self.node_dim = input_shape[2]
        self.W = self.add_weight(
            shape=(self.num_classes, self.num_nodes, self.node_dim, self.capsule_dim),
            initializer='glorot_uniform',
            trainable=True,
            name='W'
        )
        super().build(input_shape)

    def call(self, inputs):
        batch_size = tf.shape(inputs)[0]
        inputs = self.squash(inputs)
        W_perm = tf.transpose(self.W, perm=[1, 0, 2, 3])
        u_hat = tf.einsum('b i d, i j d a -> b j i a', inputs, W_perm)

        b = tf.zeros((batch_size, self.num_classes, self.num_nodes))
        for r in range(self.num_routing):
            c = tf.nn.softmax(b, axis=1)
            s = tf.reduce_sum(tf.expand_dims(c, axis=-1) * u_hat, axis=2)
            v = self.squash(s)
            if r < self.num_routing - 1:
                b += tf.reduce_sum(tf.expand_dims(v, axis=2) * u_hat, axis=-1)
        return v

    def squash(self, vector):
        vec_squared_norm = tf.reduce_sum(tf.square(vector), axis=-1, keepdims=True)
        scalar_factor = vec_squared_norm / (1.0 + vec_squared_norm)
        return scalar_factor * vector / (tf.sqrt(vec_squared_norm + tf.keras.backend.epsilon()))

    def compute_output_shape(self, input_shape):
        return (input_shape[0], self.num_classes, self.capsule_dim)

    def get_config(self):
        config = super().get_config()
        config.update({
            'num_classes': self.num_classes,
            'capsule_dim': self.capsule_dim,
            'num_routing': self.num_routing,
        })
        return config

# -------------------- GRAPH CONSTRUCTION --------------------
def build_feature_correlation_graph(X_train_np, threshold=0.2):
    corr_matrix = np.corrcoef(X_train_np, rowvar=False)
    A = (np.abs(corr_matrix) > threshold).astype(np.float32)
    np.fill_diagonal(A, 1.0)
    mask = np.where(A > 0, 0.0, -1e9)
    return mask, A   # return both mask and adjacency for GNNExplainer

# -------------------- DATA LOADING & CACHING --------------------
train_path = "/kaggle/input/datasets/amineipad/cic-iomt-dataset-2024/train_iomt.csv"
test_path  = "/kaggle/input/datasets/amineipad/cic-iomt-dataset-2024/test_iomt.csv"

cache_ready = (load_cache("X_train_scaled.npy") is not None and
               load_cache("X_test_scaled.npy") is not None and
               load_cache("y_train.npy") is not None and
               load_cache("y_test.npy") is not None and
               load_cache("label_encoder.joblib") is not None and
               load_cache("scaler.joblib") is not None and
               load_cache("feature_encoders.joblib") is not None and
               load_cache("gat_mask.npy") is not None and
               load_cache("adjacency.npy") is not None)

if cache_ready:
    print("✅ Loading preprocessed data from cache...")
    X_train_scaled = load_cache("X_train_scaled.npy")
    y_train        = load_cache("y_train.npy")
    X_test_scaled  = load_cache("X_test_scaled.npy")
    y_test         = load_cache("y_test.npy")
    label_encoder  = load_cache("label_encoder.joblib")
    scaler         = load_cache("scaler.joblib")
    gat_mask       = load_cache("gat_mask.npy")
    adjacency      = load_cache("adjacency.npy")  # for GNNExplainer
    feature_encoders = load_cache("feature_encoders.joblib")
    num_classes    = len(label_encoder.classes_)
    num_features   = X_train_scaled.shape[1]
else:
    print("🔄 Cache not found. Processing from scratch...")

    train_df = pd.read_csv(train_path, engine='pyarrow')
    test_df  = pd.read_csv(test_path,  engine='pyarrow')
    print("Train shape:", train_df.shape)
    print("Test shape:", test_df.shape)

    X_train_raw = train_df.drop("label", axis=1)
    y_train_raw = train_df["label"]
    X_test_raw  = test_df.drop("label", axis=1)
    y_test_raw  = test_df["label"]

    # Remove unseen labels from test
    train_labels_set = set(y_train_raw)
    mask = y_test_raw.isin(train_labels_set)
    if not mask.all():
        print(f"⚠️ Removing {sum(~mask)} test samples with unseen labels.")
        X_test_raw = X_test_raw[mask]
        y_test_raw = y_test_raw[mask]

    # Preprocess features (vectorized)
    def preprocess_dataframe(df, fit_encoders=None):
        df = df.copy()
        df = df.replace('-', np.nan)

        num_cols = df.select_dtypes(include=['int64', 'float64']).columns
        df[num_cols] = df[num_cols].fillna(df[num_cols].median())

        encoders = {}
        obj_cols = df.select_dtypes(include=['object']).columns
        df[obj_cols] = df[obj_cols].fillna('Unknown')
        for col in obj_cols:
            if fit_encoders is not None and col in fit_encoders:
                le = fit_encoders[col]
                df[col] = le.transform(df[col].astype(str))
            else:
                le = LabelEncoder()
                df[col] = le.fit_transform(df[col].astype(str))
                encoders[col] = le
        return df, encoders

    X_train, feature_encoders = preprocess_dataframe(X_train_raw)
    X_test, _ = preprocess_dataframe(X_test_raw, fit_encoders=feature_encoders)

    # Encode labels
    label_encoder = LabelEncoder()
    y_train = label_encoder.fit_transform(y_train_raw)
    y_test  = label_encoder.transform(y_test_raw)
    num_classes = len(label_encoder.classes_)

    # Scale features
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train).astype(np.float32)
    X_test_scaled  = scaler.transform(X_test).astype(np.float32)

    # Build correlation mask and adjacency
    gat_mask, adjacency = build_feature_correlation_graph(X_train_scaled, threshold=0.2)

    # Cache everything
    save_cache(X_train_scaled, "X_train_scaled.npy")
    save_cache(y_train, "y_train.npy")
    save_cache(X_test_scaled, "X_test_scaled.npy")
    save_cache(y_test, "y_test.npy")
    save_cache(label_encoder, "label_encoder.joblib")
    save_cache(scaler, "scaler.joblib")
    save_cache(feature_encoders, "feature_encoders.joblib")
    save_cache(gat_mask, "gat_mask.npy")
    save_cache(adjacency, "adjacency.npy")
    print("✅ Cached preprocessed data.")

num_features = X_train_scaled.shape[1]
num_classes  = len(label_encoder.classes_)
print("Classes:", num_classes, "| Features:", num_features)
feature_names = list(X_train.columns)  # for XAI

# -------------------- CLASS WEIGHTS (replaces SMOTE) --------------------
class_weights_raw = compute_class_weight('balanced',
                                         classes=np.unique(y_train),
                                         y=y_train)
CLASS_WEIGHTS = dict(enumerate(class_weights_raw))
print("Class weights:", CLASS_WEIGHTS)

# Reshape for CNN
X_train_final = X_train_scaled.reshape((-1, num_features, 1))
X_test_final  = X_test_scaled.reshape((-1, num_features, 1))

# Categorical labels (needed for margin loss)
y_train_cat = to_categorical(y_train, num_classes)
y_test_cat  = to_categorical(y_test,  num_classes)

# -------------------- MODEL CREATION FUNCTION --------------------
def create_hybrid_model():
    inputs = Input(shape=(num_features, 1))
    x = Conv1D(filters=64, kernel_size=3, padding='same', activation='relu')(inputs)
    x = Conv1D(filters=64, kernel_size=3, padding='same', activation='relu')(x)
    cnn_out = x

    x = AdaptiveTemporalAttention(reduction=16)(x)

    x = MultiHeadAttention(num_heads=4, key_dim=16, attention_axes=1)(
        query=x, key=x, value=x, attention_mask=gat_mask
    )
    x = LayerNormalization()(x + cnn_out)

    digit_caps = NodeToClassCaps(num_classes=num_classes, capsule_dim=16, num_routing=3)(x)

    def capsule_length(vector):
        return tf.sqrt(tf.reduce_sum(tf.square(vector), axis=-1) + tf.keras.backend.epsilon())

    outputs = Lambda(capsule_length)(digit_caps)
    model = Model(inputs=inputs, outputs=outputs)

    def margin_loss(y_true, y_pred):
        L = y_true * tf.square(tf.maximum(0., 0.9 - y_pred)) + \
            0.5 * (1 - y_true) * tf.square(tf.maximum(0., y_pred - 0.1))
        return tf.reduce_sum(L, axis=1)

    model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
                  loss=margin_loss, metrics=['accuracy'])
    return model


def margin_loss(y_true, y_pred):
    """Margin loss for capsule networks."""
    L = y_true * tf.square(tf.maximum(0., 0.9 - y_pred)) + \
        0.5 * (1 - y_true) * tf.square(tf.maximum(0., y_pred - 0.1))
    return tf.reduce_mean(tf.reduce_sum(L, axis=1))


def capsule_length(vector):
    """Compute the length (norm) of capsule vectors."""
    return tf.sqrt(tf.reduce_sum(tf.square(vector), axis=-1) + tf.keras.backend.epsilon())

# ==========================================================
# 3. GRAPH CONSTRUCTION
# ==========================================================
def build_feature_correlation_graph(X_train_np, threshold=0.2):
    corr_matrix = np.corrcoef(X_train_np, rowvar=False)
    A = (np.abs(corr_matrix) > threshold).astype(np.float32)
    np.fill_diagonal(A, 1.0)
    mask = np.where(A > 0, 0.0, -1e9)   # 0 = keep edge, -1e9 = block edge
    return mask


# -------------------- FEDERATED LEARNING SETUP --------------------
SEED = 42
np.random.seed(SEED)
tf.random.set_seed(SEED)
random.seed(SEED)

NUM_AGENTS = 5
NUM_ROUNDS = 15          # adjust as needed
LOCAL_EPOCHS = 5         # per round
BATCH_SIZE = 1024        # large batch
COMMUNICATION_TOPOLOGY = 'ring'
ALPHA, BETA, GAMMA = 0.4, 0.3, 0.3
TRUST_HISTORY_DECAY = 0.9

# Split training data among agents
num_train = X_train_final.shape[0]
indices = np.random.permutation(num_train)
agent_data_indices = np.array_split(indices, NUM_AGENTS)

agent_datasets = []
for idx in agent_data_indices:
    agent_X = X_train_final[idx]
    agent_y = y_train_cat[idx]
    agent_datasets.append((agent_X, agent_y))

# Initialize global model and agent models
global_model = create_hybrid_model()
global_weights = global_model.get_weights()
agent_models = [create_hybrid_model() for _ in range(NUM_AGENTS)]

# Historical trust scores (initialized equal)
hist_trust = np.ones(NUM_AGENTS) / NUM_AGENTS

best_val_acc = 0.0
best_weights_snapshot = None
history_federated = {'val_accuracy': [], 'val_loss': [], 'trust_scores': []}

# -------------------- FEDERATED TRAINING LOOP --------------------
for round_num in range(1, NUM_ROUNDS + 1):
    print(f"\n--- Trust-Aware Multi-Agent Federated Round {round_num}/{NUM_ROUNDS} ---")

    global_weights_before = [np.copy(w) for w in global_weights]

    local_weights_list = []
    local_losses = []
    update_vectors_flat = []

    for agent_id in range(NUM_AGENTS):
        print(f"  Agent {agent_id+1}/{NUM_AGENTS} local training...")
        agent_X, agent_y = agent_datasets[agent_id]

        agent_models[agent_id].set_weights(global_weights_before)

        history_local = agent_models[agent_id].fit(
            agent_X, agent_y,
            epochs=LOCAL_EPOCHS,
            batch_size=BATCH_SIZE,
            class_weight=CLASS_WEIGHTS,   # use global class weights
            verbose=0
        )

        train_loss = history_local.history['loss'][-1]
        local_losses.append(train_loss)

        agent_weights = agent_models[agent_id].get_weights()
        local_weights_list.append(agent_weights)

        # Flatten update vector
        update_flat = np.concatenate([(w - gw).flatten()
                                      for w, gw in zip(agent_weights, global_weights_before)])
        update_vectors_flat.append(update_flat)

    # ----- Compute Trust Scores -----
    avg_update = np.mean(update_vectors_flat, axis=0)
    consistency_scores = []
    for vec in update_vectors_flat:
        dot = np.dot(vec, avg_update)
        norm_vec = np.linalg.norm(vec)
        norm_avg = np.linalg.norm(avg_update)
        cos_sim = dot / (norm_vec * norm_avg + 1e-8)
        consistency_scores.append((cos_sim + 1) / 2.0)
    consistency_scores = np.array(consistency_scores)

    losses = np.array(local_losses)
    inv_losses = 1.0 / (losses + 1e-8)
    loss_scores = (inv_losses - inv_losses.min()) / (inv_losses.max() - inv_losses.min() + 1e-8)

    hist_scores = hist_trust.copy()

    raw_trust = ALPHA * consistency_scores + BETA * loss_scores + GAMMA * hist_scores
    trust_scores = raw_trust / (np.sum(raw_trust) + 1e-8)
    hist_trust = TRUST_HISTORY_DECAY * hist_trust + (1 - TRUST_HISTORY_DECAY) * trust_scores

    print(f"  Trust scores (normalized): {np.round(trust_scores, 4)}")

    # ----- Peer-to-Peer Communication (Ring) -----
    print("  Performing peer-to-peer communication (ring topology)...")
    communicated_weights = [w.copy() for w in local_weights_list]
    if COMMUNICATION_TOPOLOGY == 'ring':
        for i in range(NUM_AGENTS):
            neighbor = (i + 1) % NUM_AGENTS
            avg_pair = []
            for w_i, w_neighbor in zip(local_weights_list[i], local_weights_list[neighbor]):
                avg_pair.append((w_i + w_neighbor) / 2.0)
            communicated_weights[i] = avg_pair

    # ----- Trust-Aware Aggregation -----
    avg_weights = []
    for weight_tuple in zip(*communicated_weights):
        weighted_avg = np.average(weight_tuple, axis=0, weights=trust_scores)
        avg_weights.append(weighted_avg)

    global_weights = avg_weights
    global_model.set_weights(global_weights)

    # Evaluate on test set
    val_loss, val_acc = global_model.evaluate(X_test_final, y_test_cat, verbose=0)
    history_federated['val_loss'].append(val_loss)
    history_federated['val_accuracy'].append(val_acc)
    history_federated['trust_scores'].append(trust_scores.copy())
    print(f"  Round {round_num} - Global model val_accuracy: {val_acc:.4f}, val_loss: {val_loss:.4f}")

    if val_acc > best_val_acc:
        best_val_acc = val_acc
        best_weights_snapshot = [np.copy(w) for w in global_weights]
        global_model.save('best_fl_iomt_model.keras')
        print(f"  New best model saved (val_acc = {best_val_acc:.4f})")

print("\nTrust-Aware Federated Learning completed.")
print(f"Best validation accuracy: {best_val_acc:.4f}")

```

```python
# -------------------- OUTPUT DIRECTORY --------------------
OUTPUT_DIR = "results"
os.makedirs(OUTPUT_DIR, exist_ok=True)
print(f"📁 All outputs will be saved to: {OUTPUT_DIR}/")

# -------------------- FINAL EVALUATION (BEST MODEL) --------------------
custom_objects = {
    'NodeToClassCaps': NodeToClassCaps,
    'margin_loss': margin_loss,
    'capsule_length': capsule_length,
    'AdaptiveTemporalAttention': AdaptiveTemporalAttention
}

try:
    best_model = load_model('best_fl_iomt_model.keras', custom_objects=custom_objects)
    print("Loaded best model from disk.")
except Exception as e:
    print(f"Could not load saved model ({e}); using in-memory best weights.")
    best_model = global_model
    if best_weights_snapshot is not None:
        best_model.set_weights(best_weights_snapshot)

test_loss, test_acc = best_model.evaluate(X_test_final, y_test_cat, verbose=0)
print(f"\nBest Federated Hybrid Model Test Accuracy: {test_acc:.4f}")

y_pred = best_model.predict(X_test_final, batch_size=2048)
y_pred_classes = np.argmax(y_pred, axis=1)
y_true_classes = y_test

print("\n" + "="*60)
print("CLASSIFICATION REPORT (Federated)")
print("="*60)
print(classification_report(y_true_classes, y_pred_classes,
                            target_names=label_encoder.classes_))

# --- Confusion Matrix ---
cm = confusion_matrix(y_true_classes, y_pred_classes)
plt.figure(figsize=(12,8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=label_encoder.classes_,
            yticklabels=label_encoder.classes_)
plt.title("Confusion Matrix – Federated Hybrid Model")
plt.xlabel("Predicted"); plt.ylabel("Actual")
plt.xticks(rotation=45); plt.yticks(rotation=45)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "confusion_matrix_federated.png"), dpi=300, bbox_inches='tight')
plt.show()

# --- FL History ---
plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
plt.plot(history_federated['val_accuracy'], label='Val Accuracy')
plt.title('FL Validation Accuracy per Round')
plt.xlabel('Round'); plt.ylabel('Accuracy'); plt.legend()
plt.subplot(1,2,2)
plt.plot(history_federated['val_loss'], label='Val Loss')
plt.title('FL Validation Loss per Round')
plt.xlabel('Round'); plt.ylabel('Loss'); plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "fl_history.png"), dpi=300, bbox_inches='tight')
plt.show()

# --- Trust Scores ---
plt.figure(figsize=(10,5))
trust_array = np.array(history_federated['trust_scores'])
for i in range(NUM_AGENTS):
    plt.plot(trust_array[:, i], label=f'Agent {i+1}')
plt.title('Trust Score Evolution per Round')
plt.xlabel('Round'); plt.ylabel('Trust Score')
plt.legend(); plt.grid(True)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "trust_scores.png"), dpi=300, bbox_inches='tight')
plt.show()

# ==========================================================
# XAI — INTEGRATED GRADIENTS & GNNEXPLAINER
# ==========================================================
print("\n" + "="*60)
print("EXPLAINABLE AI (XAI)")
print("="*60)

# ---- 1. Integrated Gradients ----
def integrated_gradients(model, X_samples, target_classes=None,
                         baseline=None, steps=64, batch_size=32):
    N, F, _ = X_samples.shape
    if baseline is None:
        baseline = np.zeros((F, 1), dtype=np.float32)
    baseline = np.broadcast_to(baseline.astype(np.float32), (N, F, 1))

    if target_classes is None:
        target_classes = np.argmax(model.predict(X_samples, verbose=0), axis=1)
    target_classes = np.asarray(target_classes)

    alphas = np.linspace(0.0, 1.0, steps, dtype=np.float32)
    attributions = np.zeros((N, F, 1), dtype=np.float32)

    for s in range(0, N, batch_size):
        e = min(s + batch_size, N)
        xb = X_samples[s:e].astype(np.float32)
        bb = baseline[s:e]
        tb = target_classes[s:e]
        B = e - s

        inter = bb[None] + alphas[:, None, None, None] * (xb[None] - bb[None])
        inter = inter.reshape(steps * B, F, 1)

        grads = np.zeros_like(inter)
        tgt_flat = np.tile(tb, steps)
        for gs in range(0, inter.shape[0], batch_size):
            ge = min(gs + batch_size, inter.shape[0])
            xi = tf.convert_to_tensor(inter[gs:ge])
            with tf.GradientTape() as tape:
                tape.watch(xi)
                out = model(xi, training=False)
                picked = tf.gather(out, tgt_flat[gs:ge], batch_dims=1)
                score = tf.reduce_sum(picked)
            g = tape.gradient(score, xi)
            grads[gs:ge] = (tf.zeros_like(xi) if g is None else g).numpy()

        grads = grads.reshape(steps, B, F, 1).mean(axis=0)
        attributions[s:e] = (xb - bb) * grads

    return attributions, target_classes

# Stratified subset for explanation
N_EXPLAIN = 500
explain_idx = []
for c in np.unique(y_true_classes):
    idx_c = np.where(y_true_classes == c)[0]
    n_c = min(len(idx_c), max(1, int(N_EXPLAIN * len(idx_c) / len(y_true_classes))))
    explain_idx += list(np.random.choice(idx_c, size=n_c, replace=False))
explain_idx = np.array(explain_idx)

print(f"\nExplaining {len(explain_idx)} test samples with Integrated Gradients...")
X_explain = X_test_final[explain_idx]
y_explain = y_true_classes[explain_idx]

ig_attrs, ig_targets = integrated_gradients(best_model, X_explain,
                                            target_classes=y_explain)

global_ig = np.abs(ig_attrs).mean(axis=0).squeeze(-1)
ig_norm = global_ig / (global_ig.max() + 1e-12)
order = np.argsort(global_ig)[::-1]

print("\n" + "="*60)
print("INTEGRATED GRADIENTS — FEATURE-LEVEL ATTRIBUTION")
print("="*60)
w = max(len(f) for f in feature_names)
for f in order[:15]:
    val = ig_norm[f]
    label = 'High' if val >= 0.66 else ('Medium' if val >= 0.33 else 'Low')
    print(f"{feature_names[f]:<{w}} → {label:<6} contribution (norm={val:.2f})")

# Per-class top features
print("\nTop-5 features per class (mean |IG|):")
per_class_ig = {}
for c in range(num_classes):
    m = ig_targets == c
    if m.sum() == 0:
        continue
    per_class_ig[c] = np.abs(ig_attrs[m]).mean(axis=0).squeeze(-1)
    top = np.argsort(per_class_ig[c])[::-1][:5]
    print(f"  {label_encoder.classes_[c]:<20}: " +
          ", ".join(f"{feature_names[i]} ({per_class_ig[c][i]:.3f})" for i in top))

# --- Save IG bar plot ---
top_k = 15
plt.figure(figsize=(10,6))
plt.barh([feature_names[i] for i in order[:top_k]][::-1],
         global_ig[order[:top_k]][::-1], color='steelblue')
plt.title('Integrated Gradients — Top Global Feature Attributions')
plt.xlabel('Mean |attribution|')
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "ig_top_features.png"), dpi=300, bbox_inches='tight')
plt.show()

# ---- 2. GNNExplainer ----
A_adj = adjacency  # (F, F) binary

class GNNExplainerTF:
    BLOCK_VALUE = -10.0

    def __init__(self, model, A, base_mask, feature_names,
                 epochs=200, lr=0.05, lambda_sparsity=1e-3, lambda_entropy=1e-3):
        self.model, self.feature_names = model, feature_names
        self.A, self.A_np = tf.constant(A, tf.float32), A
        self.base_mask = tf.constant(base_mask, tf.float32)
        self.F = len(feature_names)
        self.epochs, self.lr = epochs, lr
        self.lambda_sparsity, self.lambda_entropy = lambda_sparsity, lambda_entropy

        # Decompose model
        self.gat_layer = next(l for l in model.layers if isinstance(l, MultiHeadAttention))
        self.ln_layer  = next(l for l in model.layers if isinstance(l, LayerNormalization))
        conv_layers = [l for l in model.layers if isinstance(l, Conv1D)]
        self.cnn_out_layer = conv_layers[-1].output
        self.attn_layer = next(l for l in model.layers if isinstance(l, AdaptiveTemporalAttention))
        self.pre_gat = Model(model.input, [self.cnn_out_layer, self.attn_layer.output])
        self.head = Model(self.ln_layer.output, model.output)

    def _masked_forward(self, x, edge_logits, feat_logits):
        feat_mask = tf.sigmoid(feat_logits)
        xm = x * feat_mask[None, :, None]
        cnn_out, H = self.pre_gat(xm, training=False)

        keep = tf.sigmoid(edge_logits)
        soft_block = self.A * (1.0 - keep) * self.BLOCK_VALUE
        eff_mask = self.base_mask + soft_block

        a = self.gat_layer(H, H, H, attention_mask=eff_mask, training=False)
        z = self.ln_layer(a + cnn_out, training=False)
        return self.head(z, training=False)

    def explain(self, x_np, verbose=False):
        x = tf.convert_to_tensor(x_np[None, ...], tf.float32)
        p_orig = self.model(x, training=False)
        target = int(tf.argmax(p_orig[0]))

        edge_logits = tf.Variable(np.full((self.F, self.F), 2.0, np.float32))
        feat_logits = tf.Variable(np.full((self.F,), 2.0, np.float32))
        opt = tf.keras.optimizers.Adam(self.lr)
        eps = 1e-8

        for ep in range(self.epochs):
            with tf.GradientTape() as tape:
                p_mask = self._masked_forward(x, edge_logits, feat_logits)
                fidelity = tf.reduce_sum(tf.square(p_mask - p_orig))

                keep = tf.sigmoid(edge_logits)
                fmask = tf.sigmoid(feat_logits)
                sparsity = (tf.reduce_sum(tf.abs(keep * self.A)) / tf.reduce_sum(self.A)
                            + tf.reduce_mean(tf.abs(fmask)))
                ent_e = -tf.reduce_sum(self.A * (keep * tf.math.log(keep + eps)
                                       + (1 - keep) * tf.math.log(1 - keep + eps))
                                       ) / tf.reduce_sum(self.A)
                ent_f = -tf.reduce_mean(fmask * tf.math.log(fmask + eps)
                                        + (1 - fmask) * tf.math.log(1 - fmask + eps))
                loss = (fidelity + self.lambda_sparsity * sparsity
                        + self.lambda_entropy * (ent_e + ent_f))

            grads = tape.gradient(loss, [edge_logits, feat_logits])
            opt.apply_gradients(zip(grads, [edge_logits, feat_logits]))
            if verbose and (ep + 1) % 50 == 0:
                print(f"    explainer epoch {ep+1:3d} | loss={loss.numpy():.5f}")

        return {'target': target,
                'orig_prob': float(p_orig.numpy()[0, target]),
                'edge_importance': (tf.sigmoid(edge_logits).numpy() * self.A_np).astype(np.float32),
                'node_importance': tf.sigmoid(feat_logits).numpy().astype(np.float32)}

# Run GNNExplainer on a few samples per class
GNN_PER_CLASS = 3
GNN_EPOCHS = 150  # can reduce for speed
explainer = GNNExplainerTF(best_model, A_adj, gat_mask, feature_names,
                           epochs=GNN_EPOCHS, lr=0.05,
                           lambda_sparsity=1e-3, lambda_entropy=1e-3)

gnn_indices = []
for c in range(num_classes):
    idx_c = np.where(y_explain == c)[0]
    if len(idx_c):
        gnn_indices += list(np.random.choice(idx_c, size=min(GNN_PER_CLASS, len(idx_c)), replace=False))

agg_edges = np.zeros((num_features, num_features), np.float32)
agg_nodes = np.zeros(num_features, np.float32)
per_class_edges = {c: [] for c in range(num_classes)}

print(f"\nRunning GNNExplainer on {len(gnn_indices)} samples ({GNN_EPOCHS} epochs each)...")
for n, gi in enumerate(gnn_indices):
    res = explainer.explain(X_test_final[explain_idx[gi]])
    agg_edges += res['edge_importance']
    agg_nodes += res['node_importance']
    per_class_edges[res['target']].append(res['edge_importance'])
    print(f"  [{n+1}/{len(gnn_indices)}] true={label_encoder.classes_[y_explain[gi]]}  "
          f"pred={label_encoder.classes_[res['target']]} (p={res['orig_prob']:.3f})")

agg_edges /= max(1, len(gnn_indices))
agg_nodes /= max(1, len(gnn_indices))

# Report top edges
all_edges = [(i, j, agg_edges[i, j]) for i in range(num_features)
             for j in range(i + 1, num_features) if agg_edges[i, j] > 1e-4]
all_edges.sort(key=lambda t: -t[2])

print("\n" + "="*60)
print("GNNEXPLAINER — TOP 'COMMUNICATION PATHS' (feature–feature edges)")
print("="*60)
for i, j, wt in all_edges[:12]:
    print(f"  {feature_names[i]:<{w}} ↔ {feature_names[j]:<{w}}  importance={wt:.3f}")

print("\nGNNEXPLAINER — TOP NODES (features) in explanatory subgraph:")
for f in np.argsort(agg_nodes)[::-1][:10]:
    print(f"  {feature_names[f]:<{w}} importance={agg_nodes[f]:.3f}")

# Per-class paths
print("\nPer-class explanatory subgraphs (top-3 paths):")
for c in range(num_classes):
    if not per_class_edges[c]:
        continue
    ce = np.mean(per_class_edges[c], axis=0)
    eds = sorted([(i, j, ce[i, j]) for i in range(num_features)
                  for j in range(i + 1, num_features) if ce[i, j] > 1e-4],
                 key=lambda t: -t[2])
    print(f"  Class '{label_encoder.classes_[c]}':")
    for i, j, wt in eds[:3]:
        print(f"     {feature_names[i]} ↔ {feature_names[j]}  ({wt:.3f})")

# --- Save GNN heatmap ---
plt.figure(figsize=(13, 11))
sns.heatmap(agg_edges, xticklabels=feature_names, yticklabels=feature_names,
            cmap='mako', square=True, cbar_kws={'label': 'Edge importance'})
plt.title('GNNExplainer — Aggregated Edge Importance')
plt.xticks(rotation=90); plt.yticks(rotation=0)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "gnn_edge_heatmap.png"), dpi=300, bbox_inches='tight')
plt.show()

# --- NetworkX graph (save if available) ---
try:
    import networkx as nx
    if all_edges:
        G = nx.Graph()
        for i, j, wt in all_edges[:12]:
            G.add_edge(feature_names[i], feature_names[j], weight=wt)
        node_w = {feature_names[f]: agg_nodes[f] for f in range(num_features)}
        plt.figure(figsize=(12, 9))
        pos = nx.spring_layout(G, k=1.3, seed=SEED)
        nx.draw_networkx_edges(G, pos, width=[G[u][v]['weight'] * 6 for u, v in G.edges()],
                               edge_color='steelblue', alpha=0.8)
        nx.draw_networkx_nodes(G, pos,
                               node_size=[150 + 1500 * node_w[n] for n in G.nodes()],
                               node_color='orange', alpha=0.9)
        nx.draw_networkx_labels(G, pos, font_size=9)
        plt.title('GNNExplainer — Important Subgraph')
        plt.axis('off')
        plt.tight_layout()
        plt.savefig(os.path.join(OUTPUT_DIR, "gnn_subgraph.png"), dpi=300, bbox_inches='tight')
        plt.show()
    else:
        print("No significant edges found — skipping graph plot.")
except ImportError:
    print("networkx not installed — skipping graph plot.")

# ---- 3. Combined Contribution Summary ----
gnn_norm = agg_nodes / (agg_nodes.max() + 1e-12)
combined = 0.5 * ig_norm + 0.5 * gnn_norm
combined_order = np.argsort(combined)[::-1]

print("\n" + "="*60)
print("FEATURE CONTRIBUTION SUMMARY (Integrated Gradients + GNNExplainer)")
print("="*60)
for f in combined_order[:15]:
    val = combined[f]
    label = 'High' if val >= 0.66 else ('Medium' if val >= 0.33 else 'Low')
    print(f"{feature_names[f]:<{w}} → {label:<6} contribution (score={val:.3f})")

# Save XAI artifacts into OUTPUT_DIR
pd.DataFrame({
    'feature': feature_names,
    'ig_mean_abs': global_ig,
    'ig_normalized': ig_norm,
    'gnn_node_importance': agg_nodes,
    'gnn_normalized': gnn_norm,
    'combined_score': combined
}).sort_values('combined_score', ascending=False).to_csv(
    os.path.join(OUTPUT_DIR, "xai_feature_importance_iomt.csv"), index=False)

edge_rows = [{'feature_1': feature_names[i], 'feature_2': feature_names[j], 'importance': wt}
             for i, j, wt in all_edges]
pd.DataFrame(edge_rows).to_csv(
    os.path.join(OUTPUT_DIR, "xai_gnnexplainer_edges_iomt.csv"), index=False)

print(f"\n✅ Saved CSVs and plots to folder: {OUTPUT_DIR}/")
print("All done.")
```

```python
# ==========================================================
# ADDITIONAL EVALUATION METRICS & PLOTS (with saving)
# ==========================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import (confusion_matrix, roc_curve, auc,
                             precision_recall_curve, average_precision_score,
                             roc_auc_score)
import os

# ---------- Ensure output directory exists ----------
OUTPUT_DIR = "results"
os.makedirs(OUTPUT_DIR, exist_ok=True)
print(f"📁 Saving figures to: {OUTPUT_DIR}/")

# ---------- Global font settings ----------
plt.rcParams.update({
    "font.family": "Liberation Serif",
    "font.size": 16,
    "font.weight": "bold",
    "axes.labelweight": "bold",
    "axes.titleweight": "bold"
})

# ---------- Use correct variables ----------
y_true_int = y_true_classes          # integer labels (from main)
y_pred_prob = y_pred                 # probabilities (shape [n_samples, n_classes])
y_true_onehot = y_test_cat           # one‑hot encoded true labels
class_names = label_encoder.classes_ # your encoder

n_classes = y_true_onehot.shape[1]

# ---------- 1. Overall metrics table (Macro / Micro / Weighted) ----------
cm = confusion_matrix(y_true_int, y_pred_classes)   # y_pred_classes from main
support = cm.sum(axis=1)
total_samples = cm.sum()
accuracy = np.trace(cm) / total_samples

precision = np.zeros(n_classes)
recall = np.zeros(n_classes)
f1 = np.zeros(n_classes)
specificity = np.zeros(n_classes)
tn_total = 0
fp_total = 0

for i in range(n_classes):
    tp = cm[i, i]
    fn = support[i] - tp
    fp = cm[:, i].sum() - tp
    tn = total_samples - (tp + fn + fp)

    precision[i] = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    recall[i] = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    f1[i] = 2 * precision[i] * recall[i] / (precision[i] + recall[i]) if (precision[i] + recall[i]) > 0 else 0.0
    specificity[i] = tn / (tn + fp) if (tn + fp) > 0 else 0.0

    tn_total += tn
    fp_total += fp

macro_precision = precision.mean()
macro_recall = recall.mean()
macro_f1 = f1.mean()
macro_specificity = specificity.mean()

micro_precision = accuracy
micro_recall = accuracy
micro_f1 = accuracy
micro_specificity = tn_total / (tn_total + fp_total) if (tn_total + fp_total) > 0 else 0.0

weighted_precision = np.sum(precision * support) / total_samples
weighted_recall = np.sum(recall * support) / total_samples
weighted_f1 = np.sum(f1 * support) / total_samples
weighted_specificity = np.sum(specificity * support) / total_samples

metrics_table = pd.DataFrame({
    'Accuracy': [accuracy, accuracy, accuracy],
    'Precision': [macro_precision, micro_precision, weighted_precision],
    'Recall': [macro_recall, micro_recall, weighted_recall],
    'F1-Score': [macro_f1, micro_f1, weighted_f1],
    'Specificity': [macro_specificity, micro_specificity, weighted_specificity]
}, index=['Macro', 'Micro', 'Weighted'])

print("\n" + "="*70)
print("OVERALL METRICS (Macro, Micro, Weighted)")
print("="*70)
print(metrics_table.round(4))

# ---------- 2. ROC curves (one‑vs‑rest) ----------
fpr = dict()
tpr = dict()
roc_auc = dict()

for i in range(n_classes):
    fpr[i], tpr[i], _ = roc_curve(y_true_onehot[:, i], y_pred_prob[:, i])
    roc_auc[i] = auc(fpr[i], tpr[i])

# Micro‑average
fpr["micro"], tpr["micro"], _ = roc_curve(y_true_onehot.ravel(), y_pred_prob.ravel())
roc_auc["micro"] = auc(fpr["micro"], tpr["micro"])

# Macro‑average (interpolated)
all_fpr = np.unique(np.concatenate([fpr[i] for i in range(n_classes)]))
mean_tpr = np.zeros_like(all_fpr)
for i in range(n_classes):
    mean_tpr += np.interp(all_fpr, fpr[i], tpr[i])
mean_tpr /= n_classes
fpr["macro"] = all_fpr
tpr["macro"] = mean_tpr
roc_auc["macro"] = auc(fpr["macro"], tpr["macro"])

plt.figure(figsize=(10, 8))
colors = plt.cm.tab10(np.linspace(0, 1, n_classes))
for i, color in zip(range(n_classes), colors):
    plt.plot(fpr[i], tpr[i], color=color, lw=2,
             label=f'{class_names[i]} (AUC = {roc_auc[i]:.3f})')
plt.plot(fpr["micro"], tpr["micro"],
         label=f'Micro‑average (AUC = {roc_auc["micro"]:.3f})',
         linestyle='--', linewidth=3, color='black')
plt.plot(fpr["macro"], tpr["macro"],
         label=f'Macro‑average (AUC = {roc_auc["macro"]:.3f})',
         linestyle=':', linewidth=3, color='red')
plt.plot([0, 1], [0, 1], 'k--', lw=1, alpha=0.5)
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate', fontsize=12)
plt.ylabel('True Positive Rate', fontsize=12)
plt.title('ROC Curves (One‑vs‑Rest) for Trust‑Aware Federated Hybrid Model', fontsize=14, fontweight='bold')
plt.legend(loc='lower right', fontsize=10)
plt.grid(alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "roc_curves.png"), dpi=300, bbox_inches='tight')
plt.show()

# ---------- 3. Precision‑Recall curves (one‑vs‑rest) ----------
precision = dict()
recall = dict()
average_precision = dict()

for i in range(n_classes):
    precision[i], recall[i], _ = precision_recall_curve(y_true_onehot[:, i], y_pred_prob[:, i])
    average_precision[i] = average_precision_score(y_true_onehot[:, i], y_pred_prob[:, i])

precision["micro"], recall["micro"], _ = precision_recall_curve(y_true_onehot.ravel(), y_pred_prob.ravel())
average_precision["micro"] = average_precision_score(y_true_onehot, y_pred_prob, average="micro")

plt.figure(figsize=(10, 8))
for i, color in zip(range(n_classes), colors):
    plt.plot(recall[i], precision[i], color=color, lw=2,
             label=f'{class_names[i]} (AP = {average_precision[i]:.3f})')
plt.plot(recall["micro"], precision["micro"], color='black', linestyle='--', lw=3,
         label=f'Micro‑average (AP = {average_precision["micro"]:.3f})')
plt.xlabel('Recall', fontsize=12)
plt.ylabel('Precision', fontsize=12)
plt.title('Precision‑Recall Curves (One‑vs‑Rest)', fontsize=14, fontweight='bold')
plt.legend(loc='lower left', fontsize=10)
plt.grid(alpha=0.3)
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "pr_curves.png"), dpi=300, bbox_inches='tight')
plt.show()

# ---------- 4. Threshold curves (micro‑average) ----------
prec_micro, rec_micro, thresholds_micro = precision_recall_curve(y_true_onehot.ravel(), y_pred_prob.ravel())
plt.figure(figsize=(8, 6))
plt.plot(thresholds_micro, prec_micro[:-1], label='Precision (micro)', color='blue', lw=2)
plt.plot(thresholds_micro, rec_micro[:-1], label='Recall (micro)', color='red', lw=2)
plt.xlabel('Threshold', fontsize=12)
plt.ylabel('Score', fontsize=12)
plt.title('Overall Threshold Curve (Micro‑Average)', fontsize=14, fontweight='bold')
plt.legend(loc='best', fontsize=10)
plt.grid(alpha=0.3)
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "threshold_micro.png"), dpi=300, bbox_inches='tight')
plt.show()

# ---------- 5. Threshold curves per class (one‑vs‑rest) ----------
n_cols = 2
n_rows = int(np.ceil(n_classes / n_cols))
fig, axes = plt.subplots(n_rows, n_cols, figsize=(12, 4 * n_rows))
axes = axes.flatten()

for i in range(n_classes):
    prec, rec, thresholds = precision_recall_curve(y_true_onehot[:, i], y_pred_prob[:, i])
    ax = axes[i]
    ax.plot(thresholds, prec[:-1], label='Precision', color='blue', lw=2)
    ax.plot(thresholds, rec[:-1], label='Recall', color='red', lw=2)
    ax.set_xlabel('Threshold', fontsize=10)
    ax.set_ylabel('Score', fontsize=10)
    ax.set_title(f'{class_names[i]} (One‑vs‑Rest)', fontsize=11, fontweight='bold')
    ax.legend(loc='best', fontsize=9)
    ax.grid(alpha=0.3)
    ax.set_xlim([0.0, 1.0])
    ax.set_ylim([0.0, 1.05])

for j in range(n_classes, len(axes)):
    fig.delaxes(axes[j])
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "threshold_per_class.png"), dpi=300, bbox_inches='tight')
plt.show()

# ---------- 6. Confusion matrices (raw & normalized) – separate plots ----------
# Raw counts
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=class_names, yticklabels=class_names,
            cbar=False)
plt.title('Confusion Matrix (Raw Counts)', fontsize=14, fontweight='bold')
plt.xlabel('Predicted Label', fontsize=12)
plt.ylabel('True Label', fontsize=12)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "confusion_matrix_raw.png"), dpi=300, bbox_inches='tight')
plt.show()

# Normalized (row-wise)
cm_normalized = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]
plt.figure(figsize=(10, 8))
sns.heatmap(cm_normalized, annot=True, fmt='.2%', cmap='Blues',
            xticklabels=class_names, yticklabels=class_names,
            cbar=False)
plt.title('Confusion Matrix (Normalized by True Class)', fontsize=14, fontweight='bold')
plt.xlabel('Predicted Label', fontsize=12)
plt.ylabel('True Label', fontsize=12)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "confusion_matrix_normalized.png"), dpi=300, bbox_inches='tight')
plt.show()

# ---------- 7. Misclassification heatmap ----------
cm_no_diag = cm.copy()
np.fill_diagonal(cm_no_diag, 0)
plt.figure(figsize=(10, 8))
sns.heatmap(cm_no_diag, annot=True, fmt='d', cmap='Reds',
            xticklabels=class_names, yticklabels=class_names,
            cbar_kws={'label': 'Number of Misclassifications'})
plt.title('Misclassification Heatmap (Excluding Correct Predictions)', fontsize=14, fontweight='bold')
plt.xlabel('Predicted Label', fontsize=12)
plt.ylabel('True Label', fontsize=12)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "misclassification_heatmap.png"), dpi=300, bbox_inches='tight')
plt.show()

# ---------- 8. Test set class distribution (table only, no figure) ----------
class_counts = np.sum(y_true_onehot, axis=0).astype(int)
total = class_counts.sum()
percentages = (class_counts / total) * 100
dist_df = pd.DataFrame({
    'Class': class_names,
    'Count': class_counts,
    'Percentage (%)': percentages.round(2)
})
print("\n" + "="*40)
print("TEST SET CLASS DISTRIBUTION")
print("="*40)
print(dist_df.to_string(index=False))

# ---------- 9. Per‑class ROC‑AUC and PR‑AUC bar chart ----------
roc_aucs = [roc_auc_score(y_true_onehot[:, i], y_pred_prob[:, i]) for i in range(n_classes)]
pr_aucs  = [average_precision_score(y_true_onehot[:, i], y_pred_prob[:, i]) for i in range(n_classes)]

x = np.arange(n_classes)
width = 0.35
fig, ax = plt.subplots(figsize=(12, 6))
rects1 = ax.bar(x - width/2, roc_aucs, width, label='ROC AUC', color='steelblue')
rects2 = ax.bar(x + width/2, pr_aucs, width, label='PR AUC', color='darkorange')

for i, (roc, pr) in enumerate(zip(roc_aucs, pr_aucs)):
    ax.text(x[i] - width/2, roc + 0.01, f'{roc:.4f}', ha='center', va='bottom', fontsize=9, fontweight='bold')
    ax.text(x[i] + width/2, pr + 0.01, f'{pr:.4f}', ha='center', va='bottom', fontsize=9, fontweight='bold')

ax.set_xlabel('Class', fontsize=12)
ax.set_ylabel('AUC Score', fontsize=12)
ax.set_title('Per‑Class ROC AUC and PR AUC', fontsize=14, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(class_names, rotation=45)
ax.legend()
ax.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, "per_class_auc_bars.png"), dpi=300, bbox_inches='tight')
plt.show()

print(f"\n✅ All evaluation figures saved to: {OUTPUT_DIR}/")
print("All additional evaluation plots generated.")
```
